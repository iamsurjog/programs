```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score
from sklearn.datasets import make_blobs, load_iris, make_circles
from sklearn.preprocessing import StandardScaler
```

```python
X, _ = make_circles(n_samples=300, noise=0.05, factor=0.5, random_state=42)
```

```python
plt.scatter(X[:, 0], X[:, 1])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_scaled)
y = kmeans.predict(X_scaled)
```

```python
centers = scaler.inverse_transform(kmeans.cluster_centers_)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=100, marker='X', label='Centroids')
```
